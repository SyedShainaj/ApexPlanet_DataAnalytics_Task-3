```python
!pip install dash # Install the dash library

import pandas as pd
import numpy as np
import plotly.express as px
from dash import Dash, dcc, html
from google.colab.output import eval_js
from google.colab import files

# Create dataset
np.random.seed(42)

df = pd.DataFrame({
    "Customer_ID": range(1, 801),
    "Region": np.random.choice(["North", "South", "East", "West"], 800),
    "Category": np.random.choice(["Electronics", "Fashion", "Groceries", "Home"], 800),
    "Order_Value": np.random.randint(500, 15000, 800),
    "Website_Visits": np.random.randint(10, 100, 800),
    "Purchased": np.random.choice([0, 1], 800, p=[0.30, 0.70]),
    "Churned": np.random.choice([0, 1], 800, p=[0.78, 0.22]),
    "Month": np.random.choice(["Jan", "Feb", "Mar", "Apr", "May", "Jun"], 800)
})

df.to_csv("professional_business_dashboard_dataset.csv", index=False)

# KPI calculations
total_revenue = df["Order_Value"].sum()
avg_order_value = df["Order_Value"].mean()
conversion_rate = (df["Purchased"].sum() / df["Website_Visits"].sum()) * 100
churn_rate = (df["Churned"].sum() / len(df)) * 100
total_customers = df["Customer_ID"].nunique()

# Charts
template = "plotly_dark"

region_sales = df.groupby("Region")["Order_Value"].sum().reset_index()
fig_region = px.bar(
    region_sales,
    x="Region",
    y="Order_Value",
    color="Region",
    text_auto=True,
    title="Revenue by Region",
    template=template
)

category_sales = df.groupby("Category")["Order_Value"].sum().reset_index()
fig_category = px.pie(
    category_sales,
    names="Category",
    values="Order_Value",
    title="Revenue Share by Category",
    hole=0.45,
    template=template
)

month_order = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
monthly_sales = df.groupby("Month")["Order_Value"].sum().reindex(month_order).reset_index()
fig_month = px.line(
    monthly_sales,
    x="Month",
    y="Order_Value",
    markers=True,
    title="Monthly Revenue Trend",
    template=template
)

churn = df.groupby("Region")["Churned"].mean().reset_index()
churn["Churn_Rate"] = churn["Churned"] * 100
fig_churn = px.bar(
    churn,
    x="Region",
    y="Churn_Rate",
    color="Churn_Rate",
    text_auto=".2f",
    title="Churn Rate by Region",
    template=template
)

funnel = pd.DataFrame({
    "Stage": ["Website Visits", "Purchases"],
    "Count": [df["Website_Visits"].sum(), df["Purchased"].sum()]
})
fig_funnel = px.funnel(
    funnel,
    x="Count",
    y="Stage",
    title="Customer Funnel Analysis",
    template=template
)

df["Customer_Segment"] = pd.cut(
    df["Order_Value"],
    bins=[0, 4000, 9000, 15000],
    labels=["Low Value", "Medium Value", "High Value"]
)

segment = df.groupby("Customer_Segment")["Order_Value"].sum().reset_index()
fig_segment = px.bar(
    segment,
    x="Customer_Segment",
    y="Order_Value",
    color="Customer_Segment",
    text_auto=True,
    title="Customer Segmentation",
    template=template
)

# Dashboard app
app = Dash(__name__)

card_style = {
    "background": "linear-gradient(135deg, #1f2937, #111827)",
    "padding": "22px",
    "borderRadius": "18px",
    "textAlign": "center",
    "boxShadow": "0 8px 20px rgba(0,0,0,0.35)"
}

app.layout = html.Div(
    style={
        "background": "linear-gradient(135deg, #020617, #111827, #1e1b4b)",
        "minHeight": "100vh",
        "padding": "30px",
        "fontFamily": "Arial",
        "color": "white"
    },
    children=[
        html.H1(
            "Business Intelligence Dashboard",
            style={"textAlign": "center", "fontSize": "42px", "marginBottom": "5px"}
        ),
        html.P(
            "Deep-Dive Analysis | KPIs | Customer Segmentation | Funnel Insights",
            style={"textAlign": "center", "fontSize": "18px", "color": "#cbd5e1"}
        ),

        html.Div(
            style={
                "display": "grid",
                "gridTemplateColumns": "repeat(5, 1fr)",
                "gap": "18px",
                "marginTop": "30px",
                "marginBottom": "30px"
            },
            children=[
                html.Div([html.H3("Total Revenue"), html.H2(f"₹{total_revenue:,.0f}")], style=card_style),
                html.Div([html.H3("Avg Order Value"), html.H2(f"₹{avg_order_value:,.0f}")], style=card_style),
                html.Div([html.H3("Conversion Rate"), html.H2(f"{conversion_rate:.2f}%")], style=card_style),
                html.Div([html.H3("Churn Rate"), html.H2(f"{churn_rate:.2f}%")], style=card_style),
                html.Div([html.H3("Customers"), html.H2(f"{total_customers}")], style=card_style),
            ]
        ),

        html.Div(
            style={"display": "grid", "gridTemplateColumns": "1fr 1fr", "gap": "22px"},
            children=[
                dcc.Graph(figure=fig_region),
                dcc.Graph(figure=fig_category),
                dcc.Graph(figure=fig_month),
                dcc.Graph(figure=fig_churn),
                dcc.Graph(figure=fig_funnel),
                dcc.Graph(figure=fig_segment),
            ]
        )
    ]
)

# Run dashboard in Colab
port = 8050
print("Dashboard is starting...")
print("Open this link:")
print(eval_js(f"google.colab.kernel.proxyPort({port})"))

app.run(port=port)
```
